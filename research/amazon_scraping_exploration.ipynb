```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
import numpy as np

```

```python
URL ="https://www.amazon.in/s?k=black+tshirt&crid=5KQO8EXYW0NU&sprefix=black+tsh%2Caps%2C284&ref=nb_sb_noss_2"
```

```python
HEADERS=({    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36"
    , "Accept-Language": "en-US,en;q=0.9"
})
```

```python
webpage=requests.get(URL,headers=HEADERS)
webpage
```

```python
soup=BeautifulSoup(webpage.content,"html.parser")
```

```python
links = soup.find_all("a", attrs={"class": "a-link-normal s-line-clamp-2 s-line-clamp-3-for-col-12 s-link-style a-text-normal"})
```

```python
link = links[0].get("href")
```

```python
product_list="https://www.amazon.in"+link
product_list
```

```python
new_webpage=requests.get(product_list,headers=HEADERS)
new_webpage
```

```python
new_soup=BeautifulSoup(new_webpage.content,"html.parser")
new_soup
```

```python
new_soup.find("span", attrs={"id": "productTitle"}).text.strip()
```
