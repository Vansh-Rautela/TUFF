```python
# %pip install faiss-cpu langchain-huggingface langchain-groq

from langchain_huggingface import HuggingFaceEmbeddings
from langchain.vectorstores import FAISS
from langchain_groq import ChatGroq
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain

# API Keys
import os
GROQ_API_KEY = os.environ.get("GROQ_API_KEY", "")  # never hardcode keys

class FashionAnalyzer:
    def __init__(self):
        # User's liked outfits (populated from right swipes)
        self.user_outfits = []
        
        # Initialize LLM
        self.llm = ChatGroq(model="llama-3.1-8b-instant", groq_api_key=GROQ_API_KEY)
    
    def swipe_right(self, outfit):
        """Add outfit when user swipes right"""
        self.user_outfits.append(outfit)
        print(f"✅ Added: {outfit['name']}")
    
    def analyze_fashion(self, user_question):
        """Analyze user's fashion type and recommend future products"""
        
        if not self.user_outfits:
            return "Please add some outfits to your collection first by swiping right on clothes you like!"
        
        # Build detailed context from user's collection
        outfit_details = "\n".join([
            f"• {o['name']} - Style: {o['style']}, Occasion: {o['occasion']}, Color: {o.get('color', 'N/A')}"
            for o in self.user_outfits
        ])
        
        # Create prompt template for fashion analysis
        prompt_template = """You are a professional fashion stylist and analyst. 

USER'S CLOTHING COLLECTION:
{outfit_details}

USER'S QUESTION: {question}

Based on the user's collection, provide a detailed analysis:
1. Identify their fashion type/style personality
2. Recognize patterns in their preferences (colors, styles, occasions)
3. Recommend specific products they should look for in the future
4. Explain why these recommendations match their style

Be specific, insightful, and helpful. Format your response in a clear, conversational way.
"""
        
        prompt = PromptTemplate(
            template=prompt_template,
            input_variables=["outfit_details", "question"]
        )
        
        chain = LLMChain(llm=self.llm, prompt=prompt)
        
        response = chain.run(
            outfit_details=outfit_details,
            question=user_question
        )
        
        return response


# Usage Example
if __name__ == "__main__":
    analyzer = FashionAnalyzer()
    
    # Simulate user swiping right on clothes
    print("🔄 Building your wardrobe...\n")
    
    analyzer.swipe_right({
        "name": "Casual Summer Shirt",
        "style": "casual",
        "occasion": "summer outing",
        "color": "blue"
    })
    
    analyzer.swipe_right({
        "name": "Beach Shorts",
        "style": "casual",
        "occasion": "beach",
        "color": "khaki"
    })
    
    analyzer.swipe_right({
        "name": "White Sneakers",
        "style": "casual",
        "occasion": "everyday",
        "color": "white"
    })
    
    analyzer.swipe_right({
        "name": "Denim Jacket",
        "style": "casual",
        "occasion": "outdoor",
        "color": "blue"
    })
    
    print("\n" + "="*60)
    print("💬 ASK QUESTIONS ABOUT YOUR FASHION")
    print("="*60 + "\n")
    
    # Interactive Q&A - User enters their questions
    while True:
        user_question = input("❓ Your Question (or type 'exit' to quit): ")
        
        if user_question.lower() in ['exit', 'quit', 'q']:
            print("👋 Thanks for using Fashion AI!")
            break
        
        if not user_question.strip():
            print("Please enter a question!\n")
            continue
        
        print()  # blank line
        answer = analyzer.analyze_fashion(user_question)
        print(f"🤖 Fashion AI:\n{answer}\n")
        print("-"*60 + "\n")
```
